```python
corpus = [
    "Thor ate pizza",
    "Loki is tall",
    "Loki is eating pizza"
]
```

```python
import spacy

# load english language model and create nlp object from it
nlp = spacy.load("en_core_web_sm") 

def preprocess(text):
    # remove stop words and lemmatize the text
    doc = nlp(text)
    filtered_tokens = []
    for token in doc:
        if token.is_stop or token.is_punct:
            continue
        filtered_tokens.append(token.lemma_)
    
    return " ".join(filtered_tokens) 
```

```python

preprocess("Thor ate pizza")
```

```python
corpus_processed = [
    preprocess(text) for text in corpus
]
corpus_processed
```

```python
v = CountVectorizer(ngram_range=(1,2))
v.fit(corpus_processed)
v.vocabulary_
```

```python
import pandas as pd

df = pd.read_json('News_Category_Dataset.json', lines=True)
print(df.shape)

df.head()
```

```python
cates = df.groupby('category')
print("total categories:", cates.ngroups)
print(cates.size())
```

```python
min(cates.size())
```

```python
min_sample = 1004

df_bussiness = df[df.category=="BUSINESS"].sample(min_sample, random_state=2022)
df_bussiness
```

```python
category = {c for c in df.category}
category
```

```python
df_cate = []
for catte in category:
    df_cate.append(df[df.category==catte].sample(min_sample, random_state=2022))
```

```python
df_cate
```

```python
df_balanced = pd.concat(df_cate,axis=0)
df_balanced.category.value_counts()
```

```python
df_balanced.sample()
```

```python
category_map = {
    "EDUCATION": 0,
    "IMPACT": 1,
    "POLITICS": 2,
    "SCIENCE": 3,
    "RELIGION": 4,
    "LATINO VOICES": 5,
    "HEALTHY LIVING": 6,
    "STYLE": 7,
    "ARTS & CULTURE": 8,
    "ARTS": 9,
    "PARENTS": 10,
    "BLACK VOICES": 11,
    "BUSINESS": 12,
    "THE WORLDPOST": 13,
    "GOOD NEWS": 14,
    "TRAVEL": 15,
    "MEDIA": 16,
    "TECH": 17,
    "GREEN": 18,
    "COMEDY": 19,
    "FIFTY": 20,
    "QUEER VOICES": 21,
    "WOMEN": 22,
    "CRIME": 23,
    "TASTE": 24,
    "SPORTS": 25,
    "WORLDPOST": 26,
    "WEIRD NEWS": 27,
    "WORLD NEWS": 28,
    "ENTERTAINMENT": 29,
    "COLLEGE": 30
}
```

```python
df_balanced['category_num'] = df_balanced.category.map(category_map)
df_balanced.head()
```

```python
df_balanced['text'] = df_balanced[['headline', 'short_description']].agg(' '.join, axis=1)
df_balanced.head()
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    df_balanced.text,
    df_balanced.category_num,
    test_size=0.2,
    random_state=2022,
    stratify=df_balanced.category_num
)
```

```python
print(x_train.shape)
x_train.head()
```

```python
y_test.value_counts()
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report

#1. create a pipeline object
clf = Pipeline([
     ('vectorizer_bow', CountVectorizer(ngram_range = (1, 1))),        #using the ngram_range parameter 
     ('Multi NB', MultinomialNB())         
])

#2. fit with X_train and y_train
clf.fit(x_train, y_train)


#3. get the predictions for X_test and store it in y_pred
y_pred = clf.predict(x_test)


#4. print the classfication report
print(classification_report(y_test, y_pred))
```
